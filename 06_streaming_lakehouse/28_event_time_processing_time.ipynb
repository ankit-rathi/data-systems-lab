## Build

Implement or inspect the smallest mechanism that answers the question.

```python
window=[e for e in events if 9<=e["event_time"]<=11];print([e["id"] for e in window]);print([e["id"] for e in sorted(events,key=lambda x:x["processing_time"])])
```

## Measure

Turn behavior into evidence. Avoid unsupported performance claims.

```python
print("window count:",len(window),"late event included:","c" in {e["id"] for e in window})
```

## Break

Introduce one controlled failure or adverse condition.

```python
closed=[e for e in events if e["processing_time"]<=14 and 9<=e["event_time"]<=11];print("closed before late event:",[e["id"] for e in closed])
```

## Diagnose

Explain the difference between expected and observed behavior.

```python
print("Diagnosis: event time describes when the business event happened; processing time describes when the system saw it. Windowing needs an explicit lateness policy.")
```
