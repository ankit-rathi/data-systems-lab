## Predict before you run

**Question:** What is the decision? What is the prediction time? What information is available at that time?

Write your prediction and the reason in your own words before running the code.


```
import pandas as pd
import numpy as np

rng=np.random.default_rng(42)
df=pd.DataFrame({"customer_id":range(1,101),"orders_30d":rng.poisson(3,100),"days_since_last_order":rng.integers(1,90,100)})
df["churned"]=((df.orders_30d==0)&(df.days_since_last_order>45)).astype(int)
print(df.head())
print("population:",len(df),"positive rate:",df.churned.mean())
```

## Target + baseline

```
import numpy as np
rng=np.random.default_rng(7)
score=rng.normal(0,1,200)
actual=(score>0.8).astype(int)
positive_rate=actual.mean()
print("baseline positive rate:",round(float(positive_rate),3))
assert 0<positive_rate<1
```

## Break the target boundary

```
future_signal=(score>0.2).astype(int)
print("original positives:",actual.sum(),"changed-target positives:",future_signal.sum())
assert actual.sum()!=future_signal.sum()
```
