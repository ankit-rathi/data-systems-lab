## Measure leakage impact

```
import numpy as np
rng=np.random.default_rng(8)
y=rng.integers(0,2,100)
legit=rng.normal(size=100)
leaky=y*10+rng.normal(0,0.1,100)
print("legit correlation:",round(float(np.corrcoef(legit,y)[0,1]),3))
print("leaky correlation:",round(float(np.corrcoef(leaky,y)[0,1]),3))
assert abs(np.corrcoef(leaky,y)[0,1])>abs(np.corrcoef(legit,y)[0,1])
```
