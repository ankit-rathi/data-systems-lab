## Predict before you run

**Question:** Which change is data drift, and which change is performance drift?

Write your prediction and the reason in your own words before running the code.


```
import numpy as np
rng=np.random.default_rng(42)
reference=rng.normal(0,1,1000); current=rng.normal(0.6,1,1000)
print("reference mean:",round(reference.mean(),3),"current mean:",round(current.mean(),3))
print("mean shift:",round(current.mean()-reference.mean(),3))
assert abs(current.mean()-reference.mean())>0.3
```

## Break deliberately

Introduce one controlled failure: malformed input, stale data, wrong split, missing context, unauthorized tool, or another failure relevant to this notebook. Keep the failure small and observable.


```
# Break: treat distribution shift as harmless because the model code is unchanged.
threshold=0.3
alert=abs(current.mean()-reference.mean())>threshold
print("drift alert:",alert)
assert alert
```

## Measure drift signal

```
import numpy as np
base=np.array([0.1]*100+[0.9]*100)
current=np.array([0.2]*100+[0.8]*100)
shift=abs(base.mean()-current.mean())
print("mean shift:",shift)
assert shift>0
```

## Break monitoring

```
threshold=0.05
alert=shift>threshold
print("alert:",alert)
assert alert
```
