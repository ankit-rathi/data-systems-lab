## Predict before you run

**Question:** Can we recreate the exact model that produced this prediction?

Write your prediction and the reason in your own words before running the code.


```
import hashlib, json
release={"model_version":"2026-09-30.1","code_commit":"abc123","feature_snapshot":"features-2026-09-30","parameters":{"threshold":0.5},"metric":{"auc":0.81}}
manifest=json.dumps(release,sort_keys=True).encode()
release["manifest_hash"]=hashlib.sha256(manifest).hexdigest()
print(json.dumps(release,indent=2))
assert len(release["manifest_hash"])==64
```

## Break deliberately

Introduce one controlled failure: malformed input, stale data, wrong split, missing context, unauthorized tool, or another failure relevant to this notebook. Keep the failure small and observable.


```
# Break: remove a release field and show that reproducibility evidence becomes incomplete.
broken={k:v for k,v in release.items() if k!="feature_snapshot"}
required={"model_version","code_commit","feature_snapshot","parameters"}
missing=required-set(broken)
print("missing release metadata:",missing)
assert missing=={"feature_snapshot"}
```

## Version an artifact

```
artifacts={"model-v1":{"code":"a","data":"d1"},"model-v2":{"code":"b","data":"d2"}}
active="model-v2"
print("active:",active,"bundle:",artifacts[active])
assert active in artifacts
```

## Break rollback traceability

```
target="model-v1"
rollback=target in artifacts and artifacts[target]["data"]=="d1"
print("rollback reproducible:",rollback)
assert rollback
```
