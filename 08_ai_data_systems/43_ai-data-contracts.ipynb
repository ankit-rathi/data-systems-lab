## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
contract={"version":2,"required":["request_id","context","policy_version","retrieved_at"]}
request={"request_id":"r1","context":[{"id":"refund_policy","text":"30 days","score":0.91}],"policy_version":"p7","retrieved_at":"2026-09-30"}
def validate(r,c): return [f"missing:{k}" for k in c["required"] if k not in r]
print(validate(request,contract))
```

```python
def context_quality(r):
    ctx=r.get("context",[])
    return {"has_context":bool(ctx),"max_score":max((x.get("score",0) for x in ctx),default=0),"citations":len(ctx)}
print(context_quality(request))
```

```python
# Break: stale policy and missing provenance.
bad=dict(request); bad.pop("retrieved_at"); bad["policy_version"]="p2"
print(validate(bad,contract),context_quality(bad))
```

```python
# Gate model input on both structural and policy freshness.
def admit(r,current_policy="p7"):
    errors=validate(r,contract)
    if r.get("policy_version")!=current_policy: errors.append("stale_policy")
    if context_quality(r)["max_score"]<0.8: errors.append("weak_context")
    return errors
print(admit(request)); print(admit(bad))
```
