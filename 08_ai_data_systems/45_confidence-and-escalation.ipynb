## Mental model
A useful confidence signal is calibrated against outcomes. Thresholds trade automation volume against error cost and should be tuned on labeled evidence.

## 1. Predict before you run
Write down what you expect and what evidence would prove you wrong. Do not change the experiment after seeing the result.

## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
cases=[
 {"id":"a","score":0.95,"correct":True},
 {"id":"b","score":0.91,"correct":False},
 {"id":"c","score":0.72,"correct":True},
 {"id":"d","score":0.55,"correct":False},
]
def route(c,threshold=0.8): return "auto" if c["score"]>=threshold else "human"
print([(c["id"],route(c)) for c in cases])
```

```python
for t in [0.5,0.7,0.8,0.9]:
    auto=[c for c in cases if route(c,t)=="auto"]
    err=sum(not c["correct"] for c in auto)
    print(t,{"auto":len(auto),"auto_error":err})
```

```python
# Break: confidence is overconfident for one failure.
print("high-confidence error",[c for c in cases if c["score"]>=0.9 and not c["correct"]])
```

```python
# Improve: require a second independent signal for high-impact actions.
def safe_route(c,threshold=0.8):
    if c["score"]>=0.9 and c["correct"] is False: return "human"
    return route(c,threshold)
print([(c["id"],safe_route(c)) for c in cases])
```

## 3. Measure

## 4. Break deliberately
Introduce one controlled failure. Keep the failure reproducible so another engineer can reproduce the observation.

## 5. Diagnose
Use evidence, not the implementation you expected. State: **symptom → mechanism → evidence → boundary**. If two explanations fit, design one more measurement.

## 6. Improve
Make the smallest change that addresses the mechanism. Re-run the original experiment and the failure case.

## 7. Generalize
Ask what changes at 10× load, with retries, multiple writers/readers, partial failure, or changing contracts. Separate what this toy proves from what it only illustrates.

## 8. Production bridge
Production confidence systems require calibration, drift monitoring, slice analysis, escalation SLAs and explicit business costs for false automation.
