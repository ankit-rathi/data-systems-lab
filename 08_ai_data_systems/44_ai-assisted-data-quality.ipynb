## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
rows=[{"order_id":"o1","amount":100},{"order_id":"o2","amount":-20},{"order_id":"o3","amount":0}]
profile={"rows":len(rows),"negative_amounts":sum(r["amount"]<0 for r in rows),"zero_amounts":sum(r["amount"]==0 for r in rows)}
print(profile)
```

```python
suggestions=[
 {"name":"amount_non_negative","rule":lambda r:r["amount"]>=0},
 {"name":"amount_positive","rule":lambda r:r["amount"]>0},
]
for s in suggestions: print(s["name"],sum(not s["rule"](r) for r in rows))
```

```python
# Business review rejects a rule that treats legitimate zero-value orders as invalid.
approved={"amount_non_negative"}
violations=[r for r in rows if not suggestions[0]["rule"](r)]
print("approved violations",violations)
```

```python
# Break: accept the stricter suggestion automatically.
wrong=[r for r in rows if not suggestions[1]["rule"](r)]
print("false positives introduced",wrong)
```

## 3. Measure
Do not stop at “it worked”. Record a correctness signal, a failure signal, and at least one quantitative measurement where the concept permits it.

## 4. Break deliberately
Introduce one controlled failure. Keep the failure reproducible so another engineer can reproduce the observation.

## 5. Diagnose
Use evidence, not the implementation you expected. State: **symptom → mechanism → evidence → boundary**. If two explanations fit, design one more measurement.

## 6. Improve
Make the smallest change that addresses the mechanism. Re-run the original experiment and the failure case.

## 7. Generalize
Ask what changes at 10× load, with retries, multiple writers/readers, partial failure, or changing contracts. Separate what this toy proves from what it only illustrates.

## 8. Production bridge
Production AI quality assistants need proposal/evidence separation, versioned rules, review queues, executable checks and monitoring for false positives.
