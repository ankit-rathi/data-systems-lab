## Mental model
A learning loop is controlled change: observe → label → evaluate → improve → release → monitor. Without a stable evaluation set, “improvement” is anecdotal.

## 1. Predict before you run
Write down what you expect and what evidence would prove you wrong. Do not change the experiment after seeing the result.

## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
production=[
 {"id":"r1","input":"refund 20 days","expected":"approve","observed":"approve","source":"policy"},
 {"id":"r2","input":"refund 60 days","expected":"reject","observed":"approve","source":"policy"},
 {"id":"r3","input":"refund missing receipt","expected":"human","observed":"reject","source":"policy"},
]
regression=[x for x in production if x["observed"]!=x["expected"]]
print(regression)
```

```python
# A new version changes only the known failure case.
new_observed={"r2":"reject","r3":"reject","r1":"approve"}
for x in production: print(x["id"],x["expected"],new_observed[x["id"]])
```

```python
def score(records,pred): return sum(pred[x["id"]]==x["expected"] for x in records)/len(records)
old={x["id"]:x["observed"] for x in production}
print("old",score(production,old),"new",score(production,new_observed))
```

```python
# Break the loop: remove the hard case from evaluation.
biased=[x for x in production if x["id"]!="r2"]
print("biased new score",score(biased,new_observed))
```

## 3. Measure
Do not stop at “it worked”. Record a correctness signal, a failure signal, and at least one quantitative measurement where the concept permits it.

## 4. Break deliberately
Introduce one controlled failure. Keep the failure reproducible so another engineer can reproduce the observation.

## 5. Diagnose
Use evidence, not the implementation you expected. State: **symptom → mechanism → evidence → boundary**. If two explanations fit, design one more measurement.

## 6. Improve
Make the smallest change that addresses the mechanism. Re-run the original experiment and the failure case.

## 7. Generalize
Ask what changes at 10× load, with retries, multiple writers/readers, partial failure, or changing contracts. Separate what this toy proves from what it only illustrates.

## 8. Production bridge
Production AI ops needs versioned evaluation data, release gates, feedback capture, drift/slice monitoring and a documented policy for when human labels become training or evaluation data.

## 9. Independent challenge
Add a new production failure and decide whether it belongs in the regression set, a monitoring slice, or a new capability dataset. Do not silently discard hard cases.

### Evidence to keep
regression set, before/after score, evaluation leakage demonstration, and dataset decision
