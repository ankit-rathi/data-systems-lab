## Mental model
An SLO is a target; an error budget is the remaining allowed unreliability. The budget creates a decision rule for shipping risk versus reliability work.

## 1. Predict before you run
Write down what you expect and what evidence would prove you wrong. Do not change the experiment after seeing the result.

## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
minutes_month=30*24*60
slo=0.999
budget=minutes_month*(1-slo)
print({"minutes":minutes_month,"allowed_downtime":budget})
```

```python
incidents=[{"name":"dependency","minutes":8},{"name":"bad_release","minutes":12},{"name":"region_issue","minutes":10}]
used=sum(x["minutes"] for x in incidents)
print({"used":used,"remaining":budget-used,"burn_pct":used/budget})
```

```python
# Break: a single severe incident consumes the remaining budget.
severe=budget-used+5
print("after severe incident",budget-(used+severe))
```

```python
# Improve: map budget state to a release policy.
def release_policy(remaining):
    if remaining<=0: return "freeze risky changes; reliability work"
    if remaining<budget*0.25: return "restricted changes + heightened review"
    return "normal change policy"
print(release_policy(budget-used)); print(release_policy(-1))
```
