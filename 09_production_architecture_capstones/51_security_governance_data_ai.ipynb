## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
resources={"orders":"business","payments":"restricted","customer_email":"personal"}
roles={"analyst":{"orders"},"ai_support":{"orders","customer_email"},"admin":set(resources)}
print(resources,roles)
```

```python
requests=[("ai_support","orders"),("ai_support","payments"),("analyst","customer_email")]
for role,res in requests: print(role,res,res in roles[role])
```

```python
# Add action scopes separately from data scopes.
actions={"ai_support":{"read_case","draft_reply"},"admin":{"read_case","draft_reply","issue_refund"}}
print("refund allowed for ai_support", "issue_refund" in actions["ai_support"])
```

```python
audit=[]
def authorize(role,resource,action):
    allowed=resource in roles.get(role,set()) and action in actions.get(role,set())
    audit.append({"role":role,"resource":resource,"action":action,"allowed":allowed})
    return allowed
print(authorize("ai_support","orders","draft_reply")); print(authorize("ai_support","payments","issue_refund")); print(audit)
```
