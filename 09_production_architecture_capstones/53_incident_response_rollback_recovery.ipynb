```python
current="v1"
print("code rollback",current,versions[current])
# Business verification: duplicates from v2 remain unless explicitly reconciled.
duplicate_actions=versions["v2"]["duplicate_actions"]
print("remaining side effects",duplicate_actions)
```

```python
def recovery_ready(state,expected,residual_side_effects):
    return state["error_rate"]<=expected["error_rate"] and residual_side_effects==0
print("ready before reconciliation",recovery_ready(versions[current],{"error_rate":0.02},duplicate_actions))
reconciled=dict(versions[current]); reconciled["duplicate_actions"]=0
print("ready after reconciliation",recovery_ready(reconciled,{"error_rate":0.02},0))
```

## 3. Measure
Do not stop at “it worked”. Record a correctness signal, a failure signal, and at least one quantitative measurement where the concept permits it.

## 4. Break deliberately
Introduce one controlled failure. Keep the failure reproducible so another engineer can reproduce the observation.

## 5. Diagnose
Use evidence, not the implementation you expected. State: **symptom → mechanism → evidence → boundary**. If two explanations fit, design one more measurement.

## 6. Improve
Make the smallest change that addresses the mechanism. Re-run the original experiment and the failure case.

## 7. Generalize
Ask what changes at 10× load, with retries, multiple writers/readers, partial failure, or changing contracts. Separate what this toy proves from what it only illustrates.

## 8. Production bridge
Production incident response needs detection, ownership, containment, rollback/recovery, customer communication, audit trail and a tested runbook.

## 9. Independent challenge
Inject a second failure during rollback. Decide which actions are reversible, which require reconciliation, and what evidence is needed before reopening writes.
