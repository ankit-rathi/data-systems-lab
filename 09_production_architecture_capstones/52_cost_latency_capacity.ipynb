## Mental model
Capacity planning connects workload growth to resource limits and budget. Tail latency often reflects queueing and contention that averages hide.

## 1. Predict before you run
Write down what you expect and what evidence would prove you wrong. Do not change the experiment after seeing the result.

## 2. Build
The first cell creates the smallest system that can answer the question. The later cells change one boundary at a time.

```python
workload={"avg_rps":100,"peak_rps":500,"growth":0.4,"per_worker_rps":80,"worker_monthly_cost":180}
needed=workload["peak_rps"]/workload["per_worker_rps"]
workers=math.ceil(needed) if 'math' in globals() else int(needed)+1
print({"needed":needed,"workers":workers})
```

```python
import math
workers=math.ceil(workload["peak_rps"]/workload["per_worker_rps"])
headroom=workers*workload["per_worker_rps"]/workload["peak_rps"]-1
print({"workers":workers,"headroom_pct":round(headroom*100,1),"monthly_cost":workers*workload["worker_monthly_cost"]})
```

```python
future_peak=workload["peak_rps"]*(1+workload["growth"])
future_workers=math.ceil(future_peak/workload["per_worker_rps"])
print({"future_peak":future_peak,"future_workers":future_workers})
```

```python
# Break: remove headroom and observe the capacity cliff.
for rps in [400,500,600,700]: print(rps,math.ceil(rps/workload["per_worker_rps"]))
```

## 3. Measure

## 4. Break deliberately
Introduce one controlled failure. Keep the failure reproducible so another engineer can reproduce the observation.

## 5. Diagnose
Use evidence, not the implementation you expected. State: **symptom → mechanism → evidence → boundary**. If two explanations fit, design one more measurement.

## 6. Improve
Make the smallest change that addresses the mechanism. Re-run the original experiment and the failure case.

## 7. Generalize
Ask what changes at 10× load, with retries, multiple writers/readers, partial failure, or changing contracts. Separate what this toy proves from what it only illustrates.

## 8. Production bridge
