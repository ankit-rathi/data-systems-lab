```python
gold=[True,False,True,True,False,False]
pred=[True,True,True,False,False,True]
tp=sum(g and p for g,p in zip(gold,pred)); fp=sum((not g) and p for g,p in zip(gold,pred)); fn=sum(g and (not p) for g,p in zip(gold,pred))
precision=tp/(tp+fp); recall=tp/(tp+fn)
print({"tp":tp,"fp":fp,"fn":fn,"precision":precision,"recall":recall})
```

```python
cases=[("grounded",True,True),("unsupported",False,True),("wrong_refusal",True,False)]
for name,grounded,accepted in cases: print(name,{"grounded":grounded,"accepted":accepted})
```

```python
# Break the evaluator: mark every answer as grounded.
naive=[True]*len(gold); false_grounded=sum(g is False and p for g,p in zip(gold,naive))
print("false grounded cases",false_grounded)
```
